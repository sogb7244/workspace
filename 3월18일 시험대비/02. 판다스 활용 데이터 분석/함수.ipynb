```python
#파이썬은 매개변수 사용 시 매개변수 이름을 자주 사용한다.
def hellMe(name,age,addr):
  print(f'이름은 {name}이고 나이는 {age}, 주소는 {addr}')
```

```python
hellMe('kim',20,'ulsan')
hellMe(name='kim',age=20,addr='ulsan')


```
