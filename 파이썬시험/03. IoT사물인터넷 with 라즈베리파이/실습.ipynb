```python
def get_even(lst):
    list3 = []
    for e in lst:
        if e % 2 == 0:
            list3.append(e)
    return list3

list2 = [1,2,3,4,5]
get_even(list2)
    
```

```python
def reverse(s):
    list4 = []
    for e in s:
        list4.append(e)
    list4.reverse()

    result = ''
    for i in list4:
        result += i
    return result

print(reverse("hello"))
    # return ''.join(list4)
```

```python
person = {
    'name': '김파이썬',
    'height': 1.75,
    'weight': 70
}
person['bmi'] = person['weight']/(person['height']*person['height'])
person
```

```python
bmi = person['bmi']
if bmi < 18.5:
    person['bmi_grade'] = '저체중'
elif bmi < 25.0:
    person['bmi_grade'] = '정상'
elif bmi < 30.0:
    person['bmi_grade'] = '과체중'
else:
    person['bmi_grade'] = '비만'
person
```

```python
import pandas as pd
import numpy as np

data = {
    '수강번호': ['S001', 'S002', 'S003', 'S004', 'S005', 'S006', 'S007'],
    '이름':    ['김판다', '이판다', '박판다', '정판다', '최판다', '강판다', '윤판다'],
    '수강과목': ['Python', 'Java', 'Python', 'C#', 'Java', 'Python', 'Java'],
    '점수':    [88, 72, 95, 60, 81, np.nan, 77],
    '출석일수': [20, 15, 22, 10, 18, 21, np.nan],
    '나이':    [25, 30, 22, 27, 35, 29, 24]
}

df = pd.DataFrame(data)
df.set_index('수강번호', inplace=True)
```

```python
df[['이름','수강과목']] 
```

```python
df[df['수강과목'] == 'Python'][['이름','점수']]
```
