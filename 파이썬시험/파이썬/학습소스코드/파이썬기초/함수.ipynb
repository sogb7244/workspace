```python
#파이썬은 매개변수 사용 시 매개변수 이름을 자주한다.
def tellMe(name, age, addr):
  print(f'이름은 {name}, 나이는 {age}, 주소는 {addr}')
```

```python
tellMe('kim', 20, 'ulsan')

tellMe(name='kim', age=20, addr='ulsan')
tellMe(age=20, addr='ulsan', name='kim')
```

```python
def tellMe2(name='미정', age=1, addr='알수없음'):
  print(f'이름은 {name}, 나이는 {age}, 주소는 {addr}')
```

```python
tellMe2()
tellMe2('aaa')
tellMe2(addr='울산')
```
